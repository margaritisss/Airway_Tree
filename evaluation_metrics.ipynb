```python
import open3d as o3d
import numpy as np

# Load the .ply mesh
mesh = o3d.io.read_triangle_mesh("/home/ids/gmargari-24/airway_project/DGCI/recon/generate_airways/test0000stage2_reso512_level0.0.ply")

# Calculate NM-V
nm_v = np.asarray(mesh.get_non_manifold_vertices())
print(f"NM-V count: {len(nm_v)}")
```

```python
import pyvista as pv

# Load the .ply mesh (PyVista automatically detects the format)
mesh = pv.read("/home/ids/gmargari-24/airway_project/DGCI/recon/generate_airways/test0001stage2_reso512_level0.0.ply").clean()

# Extract NM-E
nm_edges = mesh.extract_feature_edges(
    non_manifold_edges=True, 
    manifold_edges=False, 
    boundary_edges=False, 
    feature_edges=False
)
print(f"NM-E count: {nm_edges.n_cells}")
```
